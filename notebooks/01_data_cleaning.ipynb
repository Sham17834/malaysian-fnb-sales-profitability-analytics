```python
import pandas as pd
import numpy as np
```

```python
sales = pd.read_csv("../data/raw/sales_transactions.csv")
products = pd.read_csv("../data/raw/products.csv")
outlets = pd.read_csv("../data/raw/outlets.csv")
customers = pd.read_csv("../data/raw/customers.csv")
payments = pd.read_csv("../data/raw/payments.csv")
```

```python
sales = sales.drop_duplicates()
print("Rows remaining:", len(sales))
```

```python
duplicate_ids = sales[sales["transaction_id"].duplicated(keep=False)]

print("Rows with repeated transaction IDs:", len(duplicate_ids))
```

```python
for df in [sales, products, outlets, customers, payments]:
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].str.strip()
```

```python
products["product_name"] = products["product_name"].str.title()
products["category"] = products["category"].str.title()

products["category"].value_counts()
```

```python
products["category"] = products["category"].replace({"Beverages": "Beverage"})
```

```python
outlets["state"] = outlets["state"].str.title()

outlets["state"] = outlets["state"].replace(
    {"Kl": "Kuala Lumpur", "K.L.": "Kuala Lumpur"}
)

outlets["state"].value_counts()
```

```python
sales["transaction_datetime"].head(10)
```

```python
sales["transaction_datetime"] = pd.to_datetime(
    sales["transaction_datetime"], format="mixed", errors="coerce"
)
```

```python
sales["transaction_datetime"].isna().sum()
```

```python
bad_dates = sales["transaction_datetime"].isna() | ~sales[
    "transaction_datetime"
].between("2024-01-01", "2025-12-31 23:59:59")

sales[bad_dates].head()
```

```python
sales["quantity"] = sales["quantity"].replace({"one": "1", "two": "2", "three": "3"})

sales["quantity"] = pd.to_numeric(sales["quantity"], errors="coerce")
```

```python
money_cols = ["unit_price", "unit_cost"]

for col in money_cols:
    sales[col] = (
        sales[col]
        .astype("string")
        .str.replace("RM", "", regex=False)
        .str.replace(",", ".", regex=False)
        .str.strip()
    )

    sales[col] = pd.to_numeric(sales[col], errors="coerce")
```

```python
discount = sales["discount_rate"].astype("string").str.strip()

has_percent = discount.str.contains("%", regex=False, na=False)

discount = discount.str.replace("%", "", regex=False)
discount = pd.to_numeric(discount, errors="coerce")

discount.loc[has_percent] = discount.loc[has_percent] / 100

sales["discount_rate"] = discount
```

```python
print("Missing quantities:", sales["quantity"].isna().sum())
print("Invalid quantities:", (sales["quantity"] <= 0).sum())

print("Missing prices:", sales["unit_price"].isna().sum())
print("Invalid prices:", (sales["unit_price"] < 0).sum())

print("Missing costs:", sales["unit_cost"].isna().sum())
print("Invalid discounts:", (~sales["discount_rate"].between(0, 1)).sum())
```

```python
for col in ["unit_price", "unit_cost"]:
    products[col] = pd.to_numeric(products[col], errors="coerce")
```

```python
price_lookup = products.set_index("product_id")["unit_price"]
cost_lookup = products.set_index("product_id")["unit_cost"]

sales["unit_price"] = sales["unit_price"].fillna(sales["product_id"].map(price_lookup))

sales["unit_cost"] = sales["unit_cost"].fillna(sales["product_id"].map(cost_lookup))
```

```python
sales[["unit_price", "unit_cost"]].isna().sum()
```

```python
# Check whether IDs exist in master tables
bad_product = ~sales["product_id"].isin(products["product_id"])
bad_outlet = ~sales["outlet_id"].isin(outlets["outlet_id"])
bad_customer = ~sales["customer_id"].isin(customers["customer_id"])

# Check invalid numbers
bad_quantity = (
    sales["quantity"].isna() | (sales["quantity"] <= 0) | (sales["quantity"] % 1 != 0)
)

bad_price = sales["unit_price"].isna() | (sales["unit_price"] < 0)
bad_cost = sales["unit_cost"].isna() | (sales["unit_cost"] < 0)

bad_discount = ~sales["discount_rate"].between(0, 1)

# Combine all checks
invalid = (
    bad_dates
    | bad_product
    | bad_outlet
    | bad_customer
    | bad_quantity
    | bad_price
    | bad_cost
    | bad_discount
    | sales["transaction_id"].isna()
    | sales["transaction_id"].duplicated(keep=False)
)

print("Invalid sales rows:", invalid.sum())
```

```python
rejected_sales = sales[invalid].copy()
clean_sales = sales[~invalid].copy()

print("Original rows:", len(sales))
print("Valid rows:", len(clean_sales))
print("Rejected rows:", len(rejected_sales))
```

```python
clean_sales["gross_sales"] = (
    clean_sales["quantity"] * clean_sales["unit_price"]
).round(2)

clean_sales["discount_amount"] = (
    clean_sales["gross_sales"] * clean_sales["discount_rate"]
).round(2)

clean_sales["net_sales"] = (
    clean_sales["gross_sales"] - clean_sales["discount_amount"]
).round(2)

clean_sales["product_cost"] = (
    clean_sales["quantity"] * clean_sales["unit_cost"]
).round(2)

clean_sales["gross_profit"] = (
    clean_sales["net_sales"] - clean_sales["product_cost"]
).round(2)
```

```python
revenue = clean_sales["net_sales"].sum()
profit = clean_sales["gross_profit"].sum()
margin = profit / revenue * 100

print("Revenue: RM", round(revenue, 2))
print("Gross Profit: RM", round(profit, 2))
print("Gross Margin:", round(margin, 2), "%")
```

```python
payments["payment_amount"] = (
    payments["payment_amount"]
    .astype("string")
    .str.replace("RM", "", regex=False)
    .str.replace(",", ".", regex=False)
    .str.strip()
)

payments["payment_amount"] = pd.to_numeric(payments["payment_amount"], errors="coerce")
```

```python
# Keep duplicate payment IDs out of the valid set
bad_payment_id = payments["transaction_id"].isna() | payments[
    "transaction_id"
].duplicated(keep=False)

valid_payments = payments[~bad_payment_id].copy()

check = clean_sales[["transaction_id", "net_sales"]].merge(
    valid_payments, on="transaction_id", how="left", validate="one_to_one"
)

payment_issue = check["payment_amount"].isna() | (
    (check["net_sales"] - check["payment_amount"]).abs() > 0.01
)

print("Payment issues:", payment_issue.sum())
```

```python
good_ids = check.loc[~payment_issue, "transaction_id"]

payment_rejected_sales = clean_sales[
    ~clean_sales["transaction_id"].isin(good_ids)
].copy()

clean_sales = clean_sales[clean_sales["transaction_id"].isin(good_ids)].copy()

rejected_sales = pd.concat([rejected_sales, payment_rejected_sales], ignore_index=True)

clean_payments = valid_payments[valid_payments["transaction_id"].isin(good_ids)].copy()
```

```python
from pathlib import Path

output = Path("../data/processed")
output.mkdir(parents=True, exist_ok=True)
```

```python
clean_sales.to_csv(output / "sales_clean.csv", index=False)
products.to_csv(output / "products_clean.csv", index=False)
outlets.to_csv(output / "outlets_clean.csv", index=False)
customers.to_csv(output / "customers_clean.csv", index=False)
clean_payments.to_csv(output / "payments_clean.csv", index=False)
rejected_sales.to_csv(output / "rejected_sales.csv", index=False)

print("CSV files exported successfully!")
```
